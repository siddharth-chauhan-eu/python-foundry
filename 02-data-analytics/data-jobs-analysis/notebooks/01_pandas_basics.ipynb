# Pandas Basics

Exploring and analyzing data jobs using Pandas.

```python
import pandas as pd
from datasets import load_dataset
```

## Data Loading
Loaded the dataset from Hugging Face and converted it to a Pandas DataFrame.

```python
dataset = load_dataset(
    "lukebarousse/data_jobs", revision="1d815e9ce232eb27db11939c44eb048fe6d2e9ab"
)
df = dataset["train"].to_pandas()
```

## Data Cleanup
Converted the job_posted_date column from string format to datetime format.

```python
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

## Germany Data
Filtered the dataset to include only jobs located in Germany.

```python
de_jobs = df[df["job_country"] == "Germany"]
de_jobs
```

```python
job_counts = de_jobs.groupby("job_title_short").size().sort_values(ascending=False)

job_counts
```

## Salary Analysis

Calculated median, minimum, maximum salary and sample count for each job title in Germany.

```python
salary_stats = (
    de_jobs.groupby("job_title_short")["salary_year_avg"]
    .agg(["median", "min", "max", "count"])
    .sort_values(by="median", ascending=False)
)

salary_stats
```

## Key Findings

1. Senior Data Scientist roles showed the highest median salary.
2. Senior Data Engineer and Data Engineer roles were among the highest-paid positions.
3. Salary generally increased with seniority.
4. Results should be interpreted carefully because some job titles may have small sample sizes.
